# Numpy

```python
import numpy as np
```

```python
list1 = [1,2,3,4,5]
arr1 = np.array(list1)
print(arr1)
print(arr1.dtype)
```

```python
arr2 = np.zeros(5)
print(arr2)
print(arr2.dtype)
print(len(arr2))
print(arr2.size)
```

```python
list1 = [
    [10,20,30],
    [40,50,60]
]
arr1 = np.array(list1)
print(arr1)
```

```python
arr1 = np.array([
    [10,20,30],
    [40,50,60]
]
)
arr2 = np.array([
    [40,60,20],
    [30,70,10]
]
)
print(arr1 + arr2)
print(arr1 - arr2)
print(arr1 * arr2)
```

# Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
x = np.array([0,6])
y = np.array([10,30])
plt.plot(x,y)
plt.show()
```

```python
points = [3,5,90,28,45]
plt.plot(points,marker="o",linestyle="dotted",color="red")
plt.show()
```

```python
x = np.arange(-10,11,1)
y = x ** 2

plt.title("y = x ** 2")
plt.xlabel("values of x")
plt.ylabel("values of y")
plt.grid()
plt.plot(x,y)
plt.show()
```

```python
x = np.arange(-10,11,1)
y = x ** 2

plt.title("y = x ** 2")
plt.xlabel("values of x")
plt.ylabel("values of y")
plt.grid()
plt.scatter(x,y, color='orange',marker='*')
plt.show()
```

```python
x = np.arange(-10,11,1)

y1 = x ** 2
plt.subplot(3,2,1)
plt.plot(x,y1)

y2 = x ** 3
plt.subplot(3,2,2)
plt.plot(x,y2)

y3 = 2 * x
plt.subplot(3,2,3)
plt.plot(x,y3)

y4 = np.sin(x)
plt.subplot(3,2,4)
plt.plot(x,y4)

y5 = 2 - x
plt.subplot(3,2,5)
plt.plot(x,y5)

y6 = 3*x**2+5
plt.subplot(3,2,6)
plt.plot(x,y6)

plt.show()
```

```python
x = np.random.normal(150,20,200)
plt.hist(x)
plt.show()
```
