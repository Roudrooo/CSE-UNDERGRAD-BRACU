# NumPy library

```python
import numpy as np
```

Task 1

```python
a = np.random.randint(1, 101, 20)
b = np.random.randint(1, 101, 20)

print("Array a:", a)
print("Array b:", b)
```

Task 2

```python
MatA = a.reshape(5, 4)
MatB = b.reshape(5, 4)

print("Matrix A: ")
print(MatA)
print("\nMatrix B: ")
print(MatB)
```

Task 3

```python
ScalarMul = MatA * MatB

print("Scalar Multipication of Matrix A and Matrix B:", ScalarMul, sep="\n")
```

Task 4

```python
a = MatA
b = MatB.T

print("Modified Matrix B:", b, sep="\n")
print("\nAfter modification shape of Matrix B:",b.shape)

c = a @ b
print("\nMatrix Multiplication of a and b is:", c, sep="\n")
```

Task 5

```python
print("Only the column 2, 3 of Matrix C:", c[:, 1:3], sep="\n")

print("\nWhole Matrix using slicing technique:", c[:, :], sep="\n")
```

Task 6

```python
largestElem = np.max(c)
row, col = np.unravel_index(np.argmax(c), c.shape)
print("Largest element =", largestElem)
print(f"Index of the largest elemtent is = ({row}, {col})")
```

Task 7

```python
C_1D = c.flatten()

print("Original Matrix C:", c, sep="\n")
print("\nAfter flattening, C:", C_1D, sep="\n")
```

# Pandas Library

```python
import pandas as pd
```

Task 1

```python
loc = "/content/subject_scores.csv"
df = pd.read_csv(loc)
```

Task 2

```python
df.head()
```

Task 3

```python
print("Concise Summary:\n")
df.info()
print("\n\n\nDescriptive Summary:\n")
df.describe()
```

Task 4

```python
print("Null count in each column:")
print(df.isnull().sum())
```

Task 5

```python
print("Befor relacing null value:", df, sep="\n\n")

for col in df.columns:
  if df[col].dtypes != "object":
    df[col] = df[col].fillna(df[col].mean())

print("\n\nNull count in each column:")
print(df.isnull().sum())

print("\n\nAfter relacing null value:", df, sep="\n\n")
```

# Matplotlib Library

```python
import matplotlib.pyplot as plt
```

Task 1

```python
temperature_dhaka = [25,34,21,45,28,6,43,18,7,2]
humidity_dhaka = [28, 25,29,20, 26, 50, 19, 29, 52, 55]

plt.title("Scatter Plot for Temperature vs Humidity of Dhaka")
plt.xlabel("Temperature")
plt.ylabel("Humidity")
plt.grid()
plt.scatter(temperature_dhaka, humidity_dhaka, marker="*", color="red")
plt.show()
```

Task 2

```python
study_hours = [2,3,4,4, 5, 6, 7, 7, 8, 9, 9, 10, 11, 11, 12]
marks1 = [6, 10, 15, 20, 34, 44, 55, 60, 55, 67, 70, 80, 90, 99, 100]

plt.figure(figsize=(12, 8))
plt.title("Line Plot for Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid()
plt.plot(study_hours, marks1)
plt.show()
```

Task 3

```python
subjects = ['Maths', 'English', 'Science', 'Physics', 'Computer']
marks = [89, 90, 45, 78, 99]

plt.title("Horizontal Bar Graph Scores in each of the subjects")
plt.barh(subjects, marks)
plt.show()
```

Task 4

```python
colors = ["red", "green", "blue", "purple", "black"]

plt.title("Vertical Bar Graph Scores in each of the subjects")
plt.bar(subjects, marks, color = colors)
plt.show()
```

Task 5

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.title("Scatter Plot for Temperature vs Humidity of Dhaka")
plt.xlabel("Temperature")
plt.ylabel("Humidity")
plt.grid()
plt.scatter(temperature_dhaka, humidity_dhaka, marker="*", color="red")

plt.subplot(2, 2, 2)
plt.title("Line Plot for Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid()
plt.plot(study_hours, marks1)

plt.subplot(2, 2, 3)
plt.title("Horizontal Bar Graph Scores in each of the subjects")
plt.barh(subjects, marks)

plt.subplot(2, 2, 4)
plt.title("Vertical Bar Graph Scores in each of the subjects")
plt.bar(subjects, marks, color = colors)

plt.tight_layout()
plt.show()
```
